```python
import asyncio
from langchain_mcp_adapters.tools import load_mcp_tools
from langchain_openai import ChatOpenAI
from langchain.agents import create_openai_tools_agent, AgentExecutor
from langchain import hub

async def run_mcp_agent():
    # 1. 定义我们要连接的 MCP Server 地址
    # 这里假设你本地已经运行了两个 MCP Server：一个搜索，一个 Python 环境
    # 实际使用中，这些可以是 npx 启动的命令或 Docker 容器
    server_configs = {
        "tavily_search": {
            "command": "npx",
            "args": ["-y", "@modelcontextprotocol/server-tavily"],
            "env": {"TAVILY_API_KEY": "你的_API_KEY"}
        },
        "python_executor": {
            "command": "uv",
            "args": ["run", "mcp-server-python"]
        }
    }

    # 2. 核心：通过 MCP 协议自动加载工具
    # 这一步会自动把远程 Server 里的所有函数转变为 LangChain 的 Tool 对象
    mcp_tools = await load_mcp_tools(server_configs)

    # 3. 配置大脑 (LLM)
    llm = ChatOpenAI(model="gpt-4o", temperature=0)
    prompt = hub.pull("hwchase17/openai-tools-agent")

    # 4. 组装 Agent
    agent = create_openai_tools_agent(llm, mcp_tools, prompt)
    agent_executor = AgentExecutor(agent=agent, tools=mcp_tools, verbose=True)

    # 5. 执行任务
    query = "查询 2026 年 1 月 NVIDIA 的最新股价，并计算买入 100 股需要的总费用（考虑 0.1% 的手续费）"
    await agent_executor.ainvoke({"input": query})

if __name__ == "__main__":
    asyncio.run(run_mcp_agent())
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

# 1. 定义状态 (State) 🧠 —— 这是图的“共享记忆”
class State(TypedDict):
    # 我们定义一个键叫做 'name'，用来存储传递的名字
    name: str
    # 再定义一个键叫做 'greeting'，用来存储生成的结果
    greeting: str

# 2. 定义节点 (Node) 📍 —— 这是具体的“打工人”
def greeting_node(state: State):
    print("--- 正在处理问候逻辑 ---")
    # 节点接收当前状态，处理后返回更新后的状态
    return {"greeting": f"你好, {state['name']}！欢迎来到 LangGraph 的世界。"}

# 3. 构建图 (Graph) 🕸️ —— 编织逻辑网
# 初始化一个以 State 为基础的图
workflow = StateGraph(State)

# 添加节点
workflow.add_node("greeter", greeting_node)

# 设置边 (Edges)：定义流转路径
# 从起点 (START) 指向我们的问候节点
workflow.add_edge(START, "greeter")
# 从问候节点指向终点 (END)
workflow.add_edge("greeter", END)

# 4. 编译并运行 🚀
app = workflow.compile()

# 调用图：传入初始状态
input_data = {"name": "同学"}
result = app.invoke(input_data)

print(f"最终结果: {result['greeting']}")


# 
```

```python
from typing import Annotated, TypedDict, Union
from langchain_openai import ChatOpenAI
from langchain_core.messages import BaseMessage, HumanMessage
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages # 关键：用于自动合并消息列表

# 1. 定义状态 (State) 🧠
# Annotated[..., add_messages] 告诉图：当节点返回消息时，请把它附加到列表末尾，而不是替换掉列表
class State(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages]

# 2. 定义节点 (Nodes) 📍
def call_model(state: State):
    """大脑节点：负责思考和决定"""
    llm = ChatOpenAI(
    model="deepseek-chat",
    api_key=deepseek_api_key,
    base_url=deepseek_base_url,
    temperature=0
    )
    # 让模型根据目前所有的对话历史 (state['messages']) 做出反应
    response = llm.invoke(state["messages"])
    return {"messages": [response]}

def tool_node(state: State):
    """工具节点：模拟一个搜索操作"""
    print("--- 正在执行搜索工具 ---")
    # 模拟工具返回的结果
    search_result = HumanMessage(content="根据搜索结果：2026年AI领域最前沿的技术是 LangGraph 和 MCP 协议。")
    return {"messages": [search_result]}

# 3. 定义路由逻辑 (Router) 🧭
def should_continue(state: State):
    """条件判断：决定下一步去哪"""
    last_message = state["messages"][-1]
    # 如果 AI 提到了“搜索”，我们就去工具节点
    if "搜索" in last_message.content:
        return "continue"
    # 否则，我们结束流程
    return "end"

# 4. 编织复杂的图 🕸️
workflow = StateGraph(State)

# 添加节点
workflow.add_node("agent", call_model)
workflow.add_node("search_tool", tool_node)

# 设置逻辑边
workflow.add_edge(START, "agent")

# 设置【条件边】
workflow.add_conditional_edges(
    "agent",           # 起始节点
    should_continue,   # 判断函数
    {
        "continue": "search_tool", # 如果返回 continue，去 search_tool
        "end": END                 # 如果返回 end，去结束
    }
)

# 将工具节点接回 agent 形成【循环】
workflow.add_edge("search_tool", "agent")

# 5. 编译并运行
app = workflow.compile()
```
